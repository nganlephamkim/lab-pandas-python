### 6. Hãy thực hiện xử lý lần lượt tất cả dữ liệu thiếu cho các biến về điểm số còn lại.

```python
# Chọn các cột điểm số kiểu số
score_columns = [col for col in df.columns 
                 if df[col].dtype in ['int64', 'float64']]

# Thay thế dữ liệu thiếu bằng Mean
for col in score_columns:
    
    # Kiểm tra số lượng thiếu
    missing = df[col].isnull().sum()
    
    if missing > 0:
        mean_value = df[col].mean()
        df[col] = df[col].fillna(mean_value)
        
        print(f"{col}: đã thay {missing} giá trị thiếu bằng Mean = {mean_value}")
```

### 7. Tạo các biến TBM1, TBM2, TBM3 tương ứng với trung bình môn của các năm lớp 10, 11 và 12. • Công thức tính: TBM = (T2 + L + H + S + V2 + X + D + N) / 10

```python
print("\nTÍNH TRUNG BÌNH MÔN:")
# TBM lớp 10
df["TBM1"] = (df["T1"]*2 + df["L1"] + df["H1"] + df["S1"] +
              df["V1"]*2 + df["X1"] + df["D1"] + df["N1"]) / 10

# TBM lớp 11
df["TBM2"] = (df["T2"]*2 + df["L2"] + df["H2"] + df["S2"] +
              df["V2"]*2 + df["X2"] + df["D2"] + df["N2"]) / 10

# TBM lớp 12
df["TBM3"] = (df["T3"]*2 + df["L3"] + df["H3"] + df["S3"] +
              df["V3"]*2 + df["X3"] + df["D3"] + df["N3"]) / 10


print(df[["TBM1", "TBM2", "TBM3"]].head())
```

### 8. Tạo các biến xếp loại XL1, XL2 và XL3 dựa trên TBM1,TBM2 và TBM3 cho từng năm lớp 10, 11, 12 như sau: • Nhỏ hơn 5.0 xếp loại: yếu (kí hiệu là Y) • Từ 5.0 đến dưới 6.5: trung bình (kí hiệu là TB) • Từ 6.5 đến dưới 8.0: khá (kí hiệu là K) • Từ 8.0 đến dưới 9.0: giỏi (kí hiệu là G) • Từ 9.0 trở lên: xuất sắc (kí hiệu là XS)

```python
# 8. Tạo biến xếp loại XL1, XL2, XL3

print("\nXẾP LOẠI HỌC LỰC")

# Hàm xếp loại
def xep_loai(tbm):
    if tbm < 5:
        return "Y"
    elif tbm < 6.5:
        return "TB"
    elif tbm < 8:
        return "K"
    elif tbm < 9:
        return "G"
    else:
        return "XS"

# Tạo biến xếp loại
df["XL1"] = df["TBM1"].apply(xep_loai)
df["XL2"] = df["TBM2"].apply(xep_loai)
df["XL3"] = df["TBM3"].apply(xep_loai)

print(df[["TBM1", "XL1", "TBM2", "XL2", "TBM3", "XL3"]].head())
```

### 9. Tạo các biến US_TBM1, US_TBM2 và US_TBM3 để chuyển điểm trung bình các năm lớp 10, 11 và 12 từ thang điểm 10 của Việt Nam sang thang điểm 4 của Mỹ. Sử dụng phương pháp Min-Max Normalization

```python
print("\nCHUYỂN ĐỔI THANG ĐIỂM")

# Công thức Min-Max:
# X_new = ((X - Min) / (Max - Min)) * 4

# Tạo biến US_TBM1
df["US_TBM1"] = ((df["TBM1"] - df["TBM1"].min()) /
                 (df["TBM1"].max() - df["TBM1"].min())) * 4

# Tạo biến US_TBM2
df["US_TBM2"] = ((df["TBM2"] - df["TBM2"].min()) /
                 (df["TBM2"].max() - df["TBM2"].min())) * 4

# Tạo biến US_TBM3
df["US_TBM3"] = ((df["TBM3"] - df["TBM3"].min()) /
                 (df["TBM3"].max() - df["TBM3"].min())) * 4

print(df[["US_TBM1", "US_TBM2", "US_TBM3"]].head())
```

### 10. Tạo biến kết quả xét tuyển (kí hiệu là KQXT) nhằm xác định sinh viên đậu (giá trị “1”) và rớt ( giá trị “0”) vào các khối dựa trên điểm DH1, DH2 và DH3 như sau • Với khối A, A1 nếu [(DH12 + DH2 + DH3)/4] lớn hơn hoặc bằng 5.0 thì đậu, ngược lại là rớt • Với khối B nếu [(DH1 + DH22 + DH3)/4] lớn hơn hoặc bằng 5.0 thì đậu, ngược lại là rớt • Với khối khác nếu [(DH1+ DH2 + DH3)/3] lớn hơn hoặc bằng 5.0 thì đậu, ngược lại là rớt

```python
print("\nXÉT TUYỂN ĐẠI HỌC")

# Hàm xét tuyển
def xet_tuyen(row):

    khoi = row["KT"]

    # Khối A, A1
    if khoi in ["A", "A1"]:
        diem = (row["DH1"]*2 + row["DH2"] + row["DH3"]) / 4

    # Khối B
    elif khoi == "B":
        diem = (row["DH1"] + row["DH2"]*2 + row["DH3"]) / 4

    # Các khối khác
    else:
        diem = (row["DH1"] + row["DH2"] + row["DH3"]) / 3

    # Đậu / Rớt
    if diem >= 5:
        return 1
    else:
        return 0


# Tạo cột KQXT
df["KQXT"] = df.apply(xet_tuyen, axis=1)

print(df[["KT", "DH1", "DH2", "DH3", "KQXT"]].head())
```

### 3. Hãy tạo pivot-table để thống kê các giá trị count, sum, mean, median, min max, std, Q1, Q2 và Q3 của DH1 theo KT

```python
# Hàm tính Q1, Q2, Q3
def Q1(x):
    return x.quantile(0.25)

def Q2(x):
    return x.quantile(0.50)

def Q3(x):
    return x.quantile(0.75)

# Pivot-table
pivot_KT = pd.pivot_table(
    df,
    values="DH1",
    index="KT",
    aggfunc=[
        "count",
        "sum",
        "mean",
        "median",
        "min",
        "max",
        "std",
        Q1,
        Q2,
        Q3
    ]
)

print(pivot_KT)
```

## Phần 5: Mô tả dữ liệu và khảo sát dạng phân phối
### 1. Hãy mô tả và khảo sát phân phối cho biến T1 Gợi ý:

### Vẽ biểu đồ Box-Plot và xác định các 10 đại ### lượng trong biểu đồ đó

### Kiểm chứng phân phối chuẩn QQ-Plot
### Nhận xét và đánh giá về phân phối của T1

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

# Đọc dữ liệu
df = pd.read_csv("processed_dulieuxettuyendaihoc.csv")

# Thống kê mô tả
print(df["T1"].describe())

# Các đại lượng hướng tâm
print("Mean =", df["T1"].mean())
print("Median =", df["T1"].median())
print("Mode =")
print(df["T1"].mode())

# Độ phân tán
print("Variance =", df["T1"].var())
print("Standard Deviation =", df["T1"].std())
print("Min =", df["T1"].min())
print("Max =", df["T1"].max())

# Tính các đại lượng Boxplot

Q1 = df["T1"].quantile(0.25)
Q2 = df["T1"].quantile(0.50)
Q3 = df["T1"].quantile(0.75)

IQR = Q3 - Q1

Lower = Q1 - 1.5 * IQR
Upper = Q3 + 1.5 * IQR

print("Min =", df["T1"].min())
print("Q1 =", Q1)
print("Median =", Q2)
print("Q3 =", Q3)
print("Max =", df["T1"].max())
print("IQR =", IQR)
print("Lower Bound =", Lower)
print("Upper Bound =", Upper)

# Outlier
outlier = df[(df["T1"] < Lower) | (df["T1"] > Upper)]

print("Số lượng Outlier =", len(outlier))

# Boxplot
sns.boxplot(x=df["T1"])

plt.title("Boxplot của T1")

plt.show()

# So sánh Mean và Median

mean_T1 = df["T1"].mean()
median_T1 = df["T1"].median()

print("Mean =", mean_T1)
print("Median =", median_T1)

if mean_T1 > median_T1:
    print("Phân phối lệch phải")

elif mean_T1 < median_T1:
    print("Phân phối lệch trái")

else:
    print("Phân phối đối xứng")

# Histogram
df["T1"].hist(bins=10)

plt.title("Histogram của T1")
plt.xlabel("T1")
plt.ylabel("Tần số")

plt.show()

# QQ Plot
stats.probplot(df["T1"], dist="norm", plot=plt)

plt.title("QQ-Plot của T1")

plt.show()

# Độ lệch
print("Skewness =", df["T1"].skew())

# Độ nhọn
print("Kurtosis =", df["T1"].kurt())

# QQ Plot
stats.probplot(df["T1"], dist="norm", plot=plt)

plt.title("QQ-Plot của T1")

plt.show()

print("""
Nhận xét:
- Dựa vào Histogram và QQ-Plot để đánh giá phân phối.
- Nếu các điểm trên QQ-Plot gần đường thẳng thì dữ liệu gần phân phối chuẩn.
- Skewness > 0: lệch phải.
- Skewness < 0: lệch trái.
- Kurtosis lớn: phân phối nhọn.
- Kurtosis nhỏ: phân phối bẹt.
""")
```

## Vấn đề 3: Cột Weight có vấn đề về không thống nhất các đơn vị đo lường trong dữ liệu. Ta sẽ chuyển các đơn vị về thành đơn vị chuẩn “kg”

```python
# Giả lập cột Weight (chứa cả lbs và kgs) dựa trên giới tính thực tế từ file CSV
np.random.seed(42)
df['Weight'] = [f"{np.random.randint(60, 90)}kgs" if s == 'M' else f"{np.random.randint(110, 160)}lbs" for s in df['sex']]

# Vấn đề 3: Định dạng lại cân nặng, tự động quét và chuyển đổi lbs sang kgs
def clean_weight(w):
    if pd.isna(w) or str(w).lower() == 'nan': 
        return np.nan
    w_str = str(w).lower()
    num_match = re.findall(r"\d+\.?\d*", w_str)
    if not num_match: 
        return np.nan
    num = float(num_match[0])
    if 'lbs' in w_str:
        return f"{int(num / 2.2)}kgs" # Quy đổi 1kg = 2.2lbs
    return f"{int(num)}kgs"

df['Weight'] = df['Weight'].apply(clean_weight)

print("KẾT QUẢ VẤN ĐỀ 3: TOÀN BỘ CÂN NẶNG ĐÃ ĐƯỢC CHUYỂN VỀ KGS")
display(df[['Firstname', 'Weight']].head(5))
```

Còn giá trị thì là kết quả huyết áp. Chúng ta sẽ tách nội dung của cột này ra làm 3 cột sau: PulseRate : giá trị huyết áp, Sex: giới tính ( m: male, f: female) và time: thời gian (tháng-ngày) như sau:

```python
# Tạo trước các cột m0006, m0612... phân rã từ chỉ số nhịp tim thalach gốc để làm nguyên liệu cho lệnh Melt
df['m0006'] = df['thalach']
df['m0612'] = df['thalach'] - 4
df['m1218'] = df['thalach'] - 8
df['f0006'] = df['thalach'] + 3
df['f0612'] = df['thalach'] - 1
df['f1218'] = df['thalach'] + 6

# Vấn đề 8A: Gom các cột thời gian vào chung một cột tạm đặt tên là 'sex_and_time'
df_melted = pd.melt(df, id_vars=['Id', 'Age', 'Weight', 'Firstname', 'Lastname'],
                    value_vars=['m0006', 'm0612', 'm1218', 'f0006', 'f0612', 'f1218'],
                    var_name='sex_and_time', value_name='PulseRate')

print("VẤN ĐỀ 8A: BẢNG SAU KHI MELT (DỮ LIỆU ĐÃ CHUYỂN THÀNH DẠNG DỌC)")
display(df_melted[['Id', 'sex_and_time', 'PulseRate']].head(5))

# Vấn đề 8B: Tách chuỗi từ cột 'sex_and_time' để tạo ra 2 biến độc lập
# Ký tự đầu tiên đại diện cho Giới tính (M/F)
df_melted['Sex'] = df_melted['sex_and_time'].str[0].str.upper()

# Các ký tự còn lại định dạng thành chuỗi Thời gian (HH:MM)
df_melted['Time'] = df_melted['sex_and_time'].str[1:3] + ":" + df_melted['sex_and_time'].str[3:]

# Xóa bỏ cột gộp trung gian 'sex_and_time' sau khi tách thành công
df_melted.drop('sex_and_time', axis=1, inplace=True)

print("--- VẤN ĐỀ 8B: BẢNG SAU KHI TÁCH THÀNH 2 CỘT 'SEX' VÀ 'TIME' ---")
display(df_melted[['Id', 'Sex', 'Time', 'PulseRate']].head(5))
```

## Hãy khảo sát tỉ lệ dữ liệu thiếu trên biến huyết áp. Dữ liệu bị thiếu thì hãy xử lý bằng phương pháp sau • Thay thế bằng giá trị trung bình liền trước và liền sau của người đó. Nếu không được thì dùng 2) • Thay thế bằng giá trị trung bình 2 giá liền trước của người đó. Nếu không được thì dùng 3) • Thay thế bằng giá trị trung bình 2 giá liền sau của người đó. Nếu không được thì dùng 4) • Trung bình của các giá trị huyết áp của người đó. Nếu không được thì dùng 5). • Trung bình của các giá trị huyết áp của nhóm giới tính. Nếu không được thì dùng. 
## Trung bình của các giá trị dữ liệu. Nếu không được thì thay bằng mức ổn định trong y học.

```python
# Đảm bảo nhịp tim PulseRate ở dạng số
df_melted['PulseRate'] = pd.to_numeric(df_melted['PulseRate'], errors='coerce')

# 11: Định nghĩa hàm xử lý điền khuyết nhịp tim theo phân cấp ưu tiên
def hierarchical_fill(row, group_df):
    if not pd.isna(row['PulseRate']):
        return row['PulseRate']
    
    # 1. Ưu tiên lấy trung bình nhịp tim của chính bệnh nhân đó (Id)
    indiv_val = group_df[group_df['Id'] == row['Id']]['PulseRate'].mean()
    if not pd.isna(indiv_val): 
        return indiv_val
        
    # 2. Lấy trung bình theo nhóm giới tính (Sex)
    sex_val = group_df[group_df['Sex'] == row['Sex']]['PulseRate'].mean()
    if not pd.isna(sex_val): 
        return sex_val
        
    # 3. Điền giá trị ổn định y khoa mặc định
    return 72.0

df_melted['PulseRate'] = df_melted.apply(lambda r: hierarchical_fill(r, df_melted), axis=1)

print("11: ĐÃ ĐIỀN KHUYẾT TOÀN BỘ CỘT NHỊP TIM (PULSERATE)")
display(df_melted.head(5))
```

# LAB 5:NHẬN DẠNG QUẦN ÁO GIÀY DÉP THỜI TRANG VỚI BỘ DŨ LIỆU FASHION-MNIST, DÙNG THƯ VIỆN PYTORCH HUẤN LUYỆN MÔ HÌNH TRÊN GPU CỦA FIT-LAB


### 1. Import thư viện và tải dữ liệu


```python
import torch
import torch.nn as nn
import torch.optim as optim
import torchvision
import torchvision.transforms as transforms
import matplotlib.pyplot as plt
import numpy as np

transform = transforms.Compose([transforms.ToTensor()])

trainset = torchvision.datasets.FashionMNIST(root='./data', train=True,
                                             download=True, transform=transform)
trainloader = torch.utils.data.DataLoader(trainset, batch_size=64, shuffle=True)

testset = torchvision.datasets.FashionMNIST(root='./data', train=False,
                                            download=True, transform=transform)
testloader = torch.utils.data.DataLoader(testset, batch_size=64, shuffle=False)

classes = ['T-shirt/top','Trouser','Pullover','Dress','Coat',
           'Sandal','Shirt','Sneaker','Bag','Ankle boot']

print(f"Train: {len(trainset)} images, Test: {len(testset)} images")

```

### 2. Định nghĩa mô hình


```python

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(28*28, 256)
        self.fc2 = nn.Linear(256, 128)
        self.fc3 = nn.Linear(128, 10)

    def forward(self, x):
        x = x.view(-1, 28*28)
        x = torch.relu(self.fc1(x))
        x = torch.relu(self.fc2(x))
        x = self.fc3(x)
        return x

net = Net()
print(net)
```

### 3. Huấn luyện mô hình

```python

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(net.parameters(), lr=0.001)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
net.to(device)
print(f"Device: {device}")

for epoch in range(5):
    running_loss = 0.0
    for images, labels in trainloader:
        images, labels = images.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = net(images)
        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
    print(f"Epoch {epoch+1}, Loss: {running_loss/len(trainloader):.4f}")

print("Huấn luyện xong!")
```

### 4. Đánh giá mô hình


```python

correct = 0
total = 0
net.eval()
with torch.no_grad():
    for images, labels in testloader:
        images, labels = images.to(device), labels.to(device)
        outputs = net(images)
        _, predicted = torch.max(outputs.data, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

print(f"Độ chính xác trên tập test: {100 * correct / total:.2f}%")

```

### 5. Hiển thị kết quả dự đoán


```python

def imshow(img):
    npimg = img.numpy()
    plt.imshow(np.transpose(npimg, (1, 2, 0)), cmap='gray')
    plt.axis('off')
    plt.show()

dataiter = iter(testloader)
images, labels = next(dataiter)
images, labels = images.to(device), labels.to(device)
outputs = net(images)
_, predicted = torch.max(outputs, 1)

for i in range(5):
    imshow(images[i].cpu())
    print(f"Thực tế: {classes[labels[i]]} | Dự đoán: {classes[predicted[i]]}")

```

# LAB 6: XÂY DỰNG MẠNG NƠ-RON SÂU
### 1. Khai báo thư viện và cấu hình thiết bị

```python
import torch
import torch.nn as nn
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.ticker as mticker
from torch.utils.data import Dataset, DataLoader
from torchvision import datasets
from torch.optim import Adam

# Sau khi cài bản CPU, cấu hình sẽ luôn là 'cpu' để tránh lỗi DLL
device = "cpu"
print(f"Chương trình đang chạy trên thiết bị: {device}")
```

### 2. Tải bộ dữ liệu Fashion-MNIST

```python
data_folder = '~/data/FMNIST'  # Thư mục lưu trữ dữ liệu giống Lab 5

# Tải tập dữ liệu huấn luyện (Train) và tập kiểm thử (Validation)
fmnist = datasets.FashionMNIST(data_folder, download=True, train=True)
val_fmnist = datasets.FashionMNIST(data_folder, download=True, train=False)

tr_images = fmnist.data
tr_targets = fmnist.targets
val_images = val_fmnist.data
val_targets = val_fmnist.targets

print(f"Kích thước tập huấn luyện (Train): {tr_images.shape}")
print(f"Kích thước tập kiểm thử (Validation): {val_images.shape}")
```

### 3. Định nghĩa lớp Dataset và hàm khởi tạo DataLoader

```python
# 3. ĐỊNH NGHĨA LỚP ĐỐI TƯỢNG DATASET & TIỀN XỬ LÝ DỮ LIỆU
class FMNISTDataset(Dataset):
    def __init__(self, x, y):
        x = x.float() / 255.0
        x = x.view(-1, 28 * 28)
        self.x, self.y = x, y

    def __getitem__(self, ix):
        x, y = self.x[ix], self.y[ix]
        return x.to(device), y.to(device)

    def __len__(self):
        return len(self.x)

# Hàm khởi tạo DataLoader với kích thước bó batch_size = 32
def get_data():
    train = FMNISTDataset(tr_images, tr_targets)
    trn_dl = DataLoader(train, batch_size=32, shuffle=True)

    val = FMNISTDataset(val_images, val_targets)
    val_dl = DataLoader(val, batch_size=len(val_images), shuffle=False)
    return trn_dl, val_dl
```

### 4. Thiết kế kiến trúc mạng nơ-ron và hàm cấu hình mô hình

```python
# 4. THIẾT KẾ KIẾN TRÚC MÔ HÌNH VÀ BỘ TỐI ƯU HÓA
def get_model():
    # Kiến trúc mạng: Input (784) -> Hidden (1000) -> Output (10)
    model = nn.Sequential(
        nn.Linear(28 * 28, 1000),
        nn.ReLU(),
        nn.Linear(1000, 10)
    ).to(device)

    # Hàm tính toán độ lỗi ngẫu nhiên đa lớp
    loss_fn = nn.CrossEntropyLoss()
    optimizer = Adam(model.parameters(), lr=1e-3)
    return model, loss_fn, optimizer
```

### 5. Định nghĩa các hàm huấn luyện (Train) và đánh giá (Validation)

```python
# 5. XÂY DỰNG CÁC HÀM BỔ TRỢ QUÁ TRÌNH HUẤN LUYỆN & ĐÁNH GIÁ
def train_batch(x, y, model, opt, loss_fn):
    model.train()
    prediction = model(x)
    batch_loss = loss_fn(prediction, y)
    batch_loss.backward()
    opt.step()
    opt.zero_grad()
    return batch_loss.item()

@torch.no_grad()
def val_loss(x, y, model, loss_fn):
    model.eval()
    prediction = model(x)
    val_loss_val = loss_fn(prediction, y)
    return val_loss_val.item()

@torch.no_grad()
def accuracy(x, y, model):
    model.eval()
    prediction = model(x)
    max_values, argmaxes = prediction.max(-1)
    is_correct = argmaxes == y
    return is_correct.cpu().numpy().tolist()
```

### 6. Vòng lặp huấn luyện chính (Training Loop - 5 Epochs)

```python
# 6. VÒNG LẶP HUẤN LUYỆN CHÍNH (TRAINING LOOP - 5 EPOCHS)
trn_dl, val_dl = get_data()
model, loss_fn, optimizer = get_model()

train_losses, train_accuracies = [], []
val_losses, val_accuracies = [], []

print("Bắt đầu quá trình huấn luyện mô hình")
for epoch in range(5):
    train_epoch_losses, train_epoch_accuracies = [], []

    # Huấn luyện trên từng bó dữ liệu (Mini-batch)
    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        batch_loss = train_batch(x, y, model, optimizer, loss_fn)
        train_epoch_losses.append(batch_loss)
    train_epoch_loss = np.array(train_epoch_losses).mean()

    # Tính toán độ chính xác trên toàn bộ tập Train sau mỗi Epoch
    for ix, batch in enumerate(iter(trn_dl)):
        x, y = batch
        is_correct = accuracy(x, y, model)
        train_epoch_accuracies.extend(is_correct)
    train_epoch_accuracy = np.mean(train_epoch_accuracies)

    # Đánh giá hiệu năng trên tập dữ liệu kiểm thử (Validation)
    for ix, batch in enumerate(iter(val_dl)):
        x, y = batch
        val_is_correct = accuracy(x, y, model)
        validation_loss = val_loss(x, y, model, loss_fn)
        val_epoch_accuracy = np.mean(val_is_correct)

    # Lưu lại lịch sử các chỉ số qua từng Epoch để vẽ đồ thị
    train_losses.append(train_epoch_loss)
    train_accuracies.append(train_epoch_accuracy)
    val_losses.append(validation_loss)
    val_accuracies.append(val_epoch_accuracy)

    print(f"Epoch {epoch+1}/5 | Train Loss: {train_epoch_loss:.4f} | Train Acc: {train_epoch_accuracy*100:.2f}% | Val Acc: {val_epoch_accuracy*100:.2f}%")
```

### 7. Trực quan hóa kết quả (Vẽ đồ thị Loss và Accuracy)

```python
# 7. TRỰC QUAN HÓA KẾT QUẢ ĐỒ THỊ LOSS VÀ ACCURACY
epochs = np.arange(5) + 1

plt.figure(figsize=(12, 10))

# Đồ thị biểu diễn Sự biến thiên giá trị hàm Lỗi (Loss)
plt.subplot(211)
plt.plot(epochs, train_losses, 'bo-', label='Training loss')
plt.plot(epochs, val_losses, 'r.-', label='Validation loss')
plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1))
plt.title('Đồ thị hàm Loss qua các Epochs (Batch size = 32, lr = 0.001)')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.grid(True)

# Đồ thị biểu diễn Sự biến thiên của Độ chính xác (Accuracy)
plt.subplot(212)
plt.plot(epochs, train_accuracies, 'bo-', label='Training accuracy')
plt.plot(epochs, val_accuracies, 'r.-', label='Validation accuracy')
plt.gca().xaxis.set_major_locator(mticker.MultipleLocator(1))
plt.title('Đồ thị hàm Accuracy qua các Epochs')
plt.xlabel('Epochs')
plt.ylabel('Accuracy')
plt.gca().set_yticklabels(['{:.0f}%'.format(x*100) for x in plt.gca().get_yticks()])
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```

## Phần 08: Phân tích cảm xúc người dùng


```python

from nltk.sentiment import SentimentIntensityAnalyzer

# Sử dụng VADER cho sentiment analysis
sia = SentimentIntensityAnalyzer()

# Một số câu mẫu
test_sentences = [
    "This movie was absolutely fantastic! I loved every minute of it.",
    "The product is terrible, completely waste of money.",
    "It's okay, nothing special but not bad either.",
    "I am so happy with the results, exceeded my expectations!",
    "This is the worst experience I have ever had."
]

print("PHÂN TÍCH CẢM XÚC (SENTIMENT ANALYSIS):")
for sentence in test_sentences:
    scores = sia.polarity_scores(sentence)
    compound = scores['compound']
    if compound >= 0.05:
        sentiment = "Tích cực 😊"
    elif compound <= -0.05:
        sentiment = "Tiêu cực 😞"
    else:
        sentiment = "Trung tính 😐"
    print(f"\nText: {sentence}")
    print(f"Scores: {scores}")
    print(f"Kết luận: {sentiment}")


import random
from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy as nltk_accuracy

# Lấy dữ liệu movie reviews
try:
    reviews = nltk.corpus.movie_reviews
    documents = [(list(reviews.words(fileid)), category)
                 for category in reviews.categories()
                 for fileid in reviews.fileids(category)]
    random.shuffle(documents)

    # Hàm trích xuất đặc trưng
    all_words = nltk.FreqDist(w.lower() for w in reviews.words())
    word_features = list(all_words.keys())[:2000]

    def document_features(document):
        document_words = set(document)
        features = {}
        for word in word_features:
            features[f'contains({word})'] = (word in document_words)
        return features

    # Tạo featuresets
    featuresets = [(document_features(d), c) for (d, c) in documents]

    # Chia train/test
    split = int(len(featuresets) * 0.8)
    train_set, test_set = featuresets[:split], featuresets[split:]

    # Huấn luyện
    classifier = NaiveBayesClassifier.train(train_set)
    acc = nltk_accuracy(classifier, test_set)
    print(f"Độ chính xác Naive Bayes classifier: {acc:.2%}")
    print(f"\nCác đặc trưng quan trọng nhất:")
    classifier.show_most_informative_features(10)

except Exception as e:
    print(f"Lỗi: {e}")
    print("(Có thể do thiếu dữ liệu movie_reviews)")

```
